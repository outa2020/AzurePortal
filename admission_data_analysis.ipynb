# Admission Data Analysis
This notebook analyzes the admission data from AdmissionData.csv

```python
# Import required libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from datetime import datetime

# Set visualization style
sns.set_style('whitegrid')
plt.rcParams['figure.figsize'] = (12, 6)
```

## 1. Load Data

```python
# Load the admission data
df = pd.read_csv('AdmissionData.csv')

# Display basic information
print(f"Dataset shape: {df.shape}")
print(f"\nColumn names and types:")
print(df.dtypes)
print(f"\nFirst few rows:")
df.head(10)
```

## 2. Data Preprocessing

```python
# Convert Date column to datetime format
df['Date'] = pd.to_datetime(df['Date'], format='%d/%m/%Y')

# Sort by date (ascending)
df = df.sort_values('Date').reset_index(drop=True)

# Display updated dataframe
print(f"Date range: {df['Date'].min()} to {df['Date'].max()}")
df.head()
```

## 3. Exploratory Data Analysis

```python
# Statistical summary
print("Statistical Summary of Admission Count:")
print(df['AdmCount'].describe())

# Check for missing values
print(f"\nMissing values:")
print(df.isnull().sum())
```

```python
# Distribution of admission counts
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.hist(df['AdmCount'], bins=30, edgecolor='black', alpha=0.7)
plt.xlabel('Admission Count')
plt.ylabel('Frequency')
plt.title('Distribution of Admission Counts')

plt.subplot(1, 2, 2)
plt.boxplot(df['AdmCount'])
plt.ylabel('Admission Count')
plt.title('Boxplot of Admission Counts')

plt.tight_layout()
plt.show()
```

## 4. Time Series Analysis

```python
# Plot time series
plt.figure(figsize=(15, 6))
plt.plot(df['Date'], df['AdmCount'], marker='o', linestyle='-', markersize=3, alpha=0.7)
plt.xlabel('Date')
plt.ylabel('Admission Count')
plt.title('Admission Count Over Time')
plt.xticks(rotation=45)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

## 5. Weekday vs Weekend Analysis

```python
# Compare weekday vs weekend admissions
weekend_stats = df.groupby('IsWeekend')['AdmCount'].agg(['mean', 'median', 'std', 'min', 'max'])
weekend_stats.index = ['Weekday', 'Weekend']
print("Weekday vs Weekend Statistics:")
print(weekend_stats)

# Visualize
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
df.groupby('IsWeekend')['AdmCount'].mean().plot(kind='bar', color=['skyblue', 'salmon'])
plt.xticks([0, 1], ['Weekday', 'Weekend'], rotation=0)
plt.ylabel('Average Admission Count')
plt.title('Average Admissions: Weekday vs Weekend')

plt.subplot(1, 2, 2)
df.boxplot(column='AdmCount', by='IsWeekend', ax=plt.gca())
plt.xticks([1, 2], ['Weekday', 'Weekend'])
plt.xlabel('Day Type')
plt.ylabel('Admission Count')
plt.title('Admission Distribution: Weekday vs Weekend')
plt.suptitle('')  # Remove default title

plt.tight_layout()
plt.show()
```

## 6. Day of Week Analysis

```python
# Analyze by day of week
day_names = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']
dow_stats = df.groupby('DayOfWeekNum')['AdmCount'].agg(['mean', 'std', 'count'])
dow_stats.index = day_names

print("Statistics by Day of Week:")
print(dow_stats)

# Visualize
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
dow_stats['mean'].plot(kind='bar', color='steelblue', alpha=0.7)
plt.xlabel('Day of Week')
plt.ylabel('Average Admission Count')
plt.title('Average Admissions by Day of Week')
plt.xticks(rotation=45, ha='right')

plt.subplot(1, 2, 2)
df.boxplot(column='AdmCount', by='DayOfWeekNum', ax=plt.gca())
plt.xticks(range(1, 8), day_names, rotation=45, ha='right')
plt.xlabel('Day of Week')
plt.ylabel('Admission Count')
plt.title('Admission Distribution by Day of Week')
plt.suptitle('')  # Remove default title

plt.tight_layout()
plt.show()
```

## 7. Monthly Analysis

```python
# Analyze by month
month_names = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']
monthly_stats = df.groupby('MonthNum')['AdmCount'].agg(['mean', 'std', 'sum', 'count'])
monthly_stats.index = [month_names[i-1] for i in monthly_stats.index]

print("Statistics by Month:")
print(monthly_stats)

# Visualize
plt.figure(figsize=(15, 5))

plt.subplot(1, 2, 1)
monthly_stats['mean'].plot(kind='bar', color='forestgreen', alpha=0.7)
plt.xlabel('Month')
plt.ylabel('Average Admission Count')
plt.title('Average Admissions by Month')
plt.xticks(rotation=45)

plt.subplot(1, 2, 2)
monthly_stats['sum'].plot(kind='bar', color='orange', alpha=0.7)
plt.xlabel('Month')
plt.ylabel('Total Admissions')
plt.title('Total Admissions by Month')
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

## 8. Correlation Analysis

```python
# Calculate correlations
correlation_matrix = df[['DayOfWeekNum', 'IsWeekend', 'MonthNum', 'AdmCount']].corr()

print("Correlation Matrix:")
print(correlation_matrix)

# Visualize correlation matrix
plt.figure(figsize=(8, 6))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', center=0, 
            square=True, linewidths=1, cbar_kws={"shrink": 0.8})
plt.title('Correlation Matrix')
plt.tight_layout()
plt.show()
```

## 9. Rolling Averages

```python
# Calculate rolling averages
df['MA_7'] = df['AdmCount'].rolling(window=7).mean()
df['MA_30'] = df['AdmCount'].rolling(window=30).mean()

# Plot
plt.figure(figsize=(15, 6))
plt.plot(df['Date'], df['AdmCount'], label='Actual', alpha=0.5, linewidth=1)
plt.plot(df['Date'], df['MA_7'], label='7-Day Moving Average', linewidth=2)
plt.plot(df['Date'], df['MA_30'], label='30-Day Moving Average', linewidth=2)
plt.xlabel('Date')
plt.ylabel('Admission Count')
plt.title('Admission Count with Moving Averages')
plt.legend()
plt.xticks(rotation=45)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

## 10. Summary Statistics

```python
# Overall summary
print("=" * 60)
print("ADMISSION DATA SUMMARY")
print("=" * 60)
print(f"Total Records: {len(df)}")
print(f"Date Range: {df['Date'].min().strftime('%d/%m/%Y')} to {df['Date'].max().strftime('%d/%m/%Y')}")
print(f"\nTotal Admissions: {df['AdmCount'].sum()}")
print(f"Average Daily Admissions: {df['AdmCount'].mean():.2f}")
print(f"Median Daily Admissions: {df['AdmCount'].median():.2f}")
print(f"Std Dev: {df['AdmCount'].std():.2f}")
print(f"\nMinimum: {df['AdmCount'].min()} on {df[df['AdmCount'] == df['AdmCount'].min()]['Date'].values[0]}")
print(f"Maximum: {df['AdmCount'].max()} on {df[df['AdmCount'] == df['AdmCount'].max()]['Date'].values[0]}")
print(f"\nWeekday Average: {df[df['IsWeekend'] == 0]['AdmCount'].mean():.2f}")
print(f"Weekend Average: {df[df['IsWeekend'] == 1]['AdmCount'].mean():.2f}")
print("=" * 60)
```
