“Failed to create endpoint request … Could not find datastore: azureml.”
Trace ID: 9c6c3bbd-b711-4525-ac6b-5f9df54f608e

Quick fix (Studio UI)


Check existing datastores

In Azure ML Studio → Manage → Data → Datastores, confirm you have workspaceblobstore (default) and whether azureml is registered.
By design, workspaceblobstore is created when the workspace is provisioned and set as default; if it’s missing or not a blob type, jobs/deployments can fail. [docs.azure.cn]



Register a datastore named azureml (if your deployment expects that name)

+ New datastore → Azure Blob Storage
Name: azureml
Storage account & container: point to the blob container you want (can be the container named azureml).
Choose your authentication (Account key, SAS, or identity-based).
This resolves the “could not find datastore azureml” error because AML is looking for a registered datastore by that exact name. [learn.microsoft.com]



(If workspaceblobstore is missing)

Create a blob datastore and set it as default, or ensure the workspace’s default storage is a valid Azure Blob and not deleted/corrupted. A missing or non-blob default store can cause uploads/deploys to fail. [stackoverflow.com]

```python
# List datastores
az ml datastore list -g <rg> -w <workspace> -o table
```

```python
# If azureml isn’t listed, create it (Blob example):

# Create a blob datastore named 'azureml'
cat > blobstore.yml <<'YAML'
$schema: https://azuremlschemas.azureedge.net/latest/datastore.schema.json
name: azureml
type: azure_blob
account_name: <storageAccountName>
container_name: <containerName>   # e.g., azureml
credentials:
  credential_type: account_key    # or sas / none for MSI
YAML

az ml datastore create \
  --file blobstore.yml \
  -g <rg> -w <workspace>

```
